# AULA SOBRE RECURSIVIDADE

```A função chama ela mesma```

````python
def fatorial(n):
    if n > 1:
        return n * fatorial(n - 1)
    else:
        return 1
````

````python
f = fatorial(5)
````

````python
print(f)
````

````python
def fibonacci(n):
    if n < 3:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)
````

````python
fib = fibonacci(5)
````

````python
print(fib)
````

````python
def soma_digito(n):
    if n < 0:
        return "Valor inválido"
    elif n <= 10:
        return n
    else:   
        return n % 10 + soma_digito(n // 10)
        
````

````python
soma_digito(123)
````
