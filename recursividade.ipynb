# AULA SOBRE RECURSIVIDADE

```A função chama ela mesma```

````python
def fatorial(n):
    if n > 1:
        return n * fatorial(n - 1)
    else:
        return 1
````

````python
f = fatorial(5)
````

````python
print(f)
````

````python
def fibonacci(n):
    if n < 3:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)
````

````python
fib = fibonacci(5)
````

````python
print(fib)
````

````python
def soma_digito(n):
    if n < 0:
        return "Valor inválido"
    elif n <= 10:
        return n
    else:   
        return n % 10 + soma_digito(n // 10)
        
````

````python
soma_digito(123)
````

# Exercicios

Exercício 1 — Caso base e passo recursivo

Resposta: Alternativa D.

Toda função recursiva deve possuir um caso base e chamar a si mesma com um problema menor ou mais simples.

Exercício 2 — Complexidade do Fibonacci recursivo

Resposta: Alternativa C — O(2^n).

A versão recursiva simples do Fibonacci possui crescimento exponencial, pois cada chamada pode gerar duas novas chamadas recursivas, repetindo cálculos já realizados.***

````python
"""Exercício 3 — Soma dos dígitos"""

def soma_digitos(n):
    if n < 10:
        return n

    ultimo_digito = n % 10
    resto_do_numero = n // 10

    return ultimo_digito + soma_digitos(resto_do_numero)


print(soma_digitos(123))
print(soma_digitos(5072))
````

````python
"""Exercício 4 — Verificação de palíndromo"""

def palindromo(s):
    s = s.lower().replace(" ", "")

    if len(s) <= 1:
        return True

    if s[0] != s[-1]:
        return False

    return palindromo(s[1:-1])


print(palindromo("A man a plan a canal Panama"))
print(palindromo("Socorram me subi no onibus em Marrocos"))
print(palindromo("hello"))
````

````python
"""Exercício 5 — Convertendo uma soma iterativa em recursiva"""

def soma_recursiva(n):
    if n == 1:
        return 1

    return n + soma_recursiva(n - 1)


print(soma_recursiva(5))
````

````python
"""Projeto 1 — Analisando uma estrutura organizacional hierárquica"""

def contar_funcionarios(departamento):
    total = departamento.get("funcionarios", 0)
    subdepartamentos = departamento.get("subdepartamentos", [])

    for subdepartamento in subdepartamentos:
        total += contar_funcionarios(subdepartamento)

    return total


estrutura = {
    "nome": "Diretoria",
    "funcionarios": 10,
    "subdepartamentos": [
        {
            "nome": "Financeiro",
            "funcionarios": 5
        },
        {
            "nome": "Tecnologia",
            "funcionarios": 8,
            "subdepartamentos": [
                {"nome": "Dados", "funcionarios": 3},
                {"nome": "Infraestrutura", "funcionarios": 2}
            ]
        }
    ]
}

print(f"Total de funcionários: {contar_funcionarios(estrutura)}")
````

````python
"""Projeto 2 — Buscando arquivos por extensão"""

import os


def encontrar_arquivos(diretorio, extensao):
    arquivos_encontrados = []

    try:
        for item in os.listdir(diretorio):
            caminho_completo = os.path.join(diretorio, item)

            if os.path.isfile(caminho_completo):
                if caminho_completo.lower().endswith(extensao.lower()):
                    arquivos_encontrados.append(caminho_completo)

            elif os.path.isdir(caminho_completo):
                encontrados_no_subdiretorio = encontrar_arquivos(
                    caminho_completo, extensao
                )
                arquivos_encontrados.extend(encontrados_no_subdiretorio)

    except FileNotFoundError:
        print(f"Erro: diretório '{diretorio}' não encontrado.")

    return arquivos_encontrados


diretorio_exemplo = "./dados"
extensao_procurada = ".csv"

arquivos = encontrar_arquivos(diretorio_exemplo, extensao_procurada)

for arquivo in arquivos:
    print(arquivo)
````
