Exercício 2 — Identificando o Merge Sort

Resposta: Alternativa E — O(n × log n).

O algoritmo descrito é o Merge Sort, que utiliza a estratégia de dividir para conquistar. Ele divide a lista em duas partes, ordena cada parte recursivamente e depois realiza a intercalação.

Sua complexidade é O(n log n), pois existem aproximadamente log n níveis de divisão e cada nível processa n elementos.

Exercício 3 — Comparando Merge, Quick, Selection e Insertion

Resposta: Alternativa D — I, II, III e IV.

Todas as afirmações são verdadeiras:

I — Merge Sort é estável e possui complexidade O(n log n) no pior caso.

II — Quick Sort possui O(n²) no pior caso e O(n log n) em média.

III — Selection Sort possui O(n²) no pior caso e não é estável na implementação padrão.

IV — Insertion Sort possui O(n²) no pior caso, mas pode se aproximar de O(n) em listas quase ordenadas.

Exercício 4 — Bubble Sort e Quick Sort em vetor decrescente

Resposta: Alternativa C — I, II e IV.

I — Verdadeiro. O Bubble Sort apresenta complexidade O(n²) nesse cenário.

II — Verdadeiro. O Quick Sort pode apresentar O(n²) quando o pivô é mal escolhido.

III — Falso. O Quick Sort tende a ser mais eficiente que o Bubble Sort em grandes conjuntos.

IV — Verdadeiro. Com uma boa escolha de pivô, o Quick Sort pode apresentar desempenho O(n log n).

Exercício 5 — Melhor algoritmo para vetor quase ordenado

O algoritmo mais indicado é o Insertion Sort.

Ele aproveita a ordenação que já existe no vetor, realizando poucos deslocamentos quando existem apenas alguns elementos fora de posição.

Em listas já ordenadas, sua complexidade pode chegar a O(n), sendo também eficiente para listas quase ordenadas.

Exercício 6 — Estabilidade na ordenação

Os algoritmos estáveis estudados são:

Bubble Sort;

Insertion Sort;

Merge Sort.

Estabilidade significa que, quando dois elementos possuem o mesmo valor de comparação, a ordem relativa entre eles é preservada após a ordenação.

Um exemplo é a organização de clientes por cidade. Se dois clientes pertencem à mesma cidade, um algoritmo estável mantém a ordem de cadastro que eles já possuíam.

Exercício 8 — Ordenando milhões de elementos

Os algoritmos mais indicados são o Quicksort ou Merge Sort, pois apresentam complexidade O(n log n).

O Quicksort possui bom desempenho médio e utiliza pouca memória auxiliar. Já o Merge Sort garante O(n log n) mesmo no pior caso, é estável, mas necessita de memória auxiliar O(n).

A escolha depende dos requisitos da aplicação.

```python
#Projeto 1 — Análise de Vendas Ordenadas

vendas = [120, 85, 90, 100, 75, 130, 95, 110, 80, 140, 70, 105]


def particao_decrescente(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] >= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def quick_sort_decrescente(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1

    if low < high:
        pi = particao_decrescente(arr, low, high)

        quick_sort_decrescente(arr, low, pi - 1)
        quick_sort_decrescente(arr, pi + 1, high)

    return arr


vendas_ordenadas = quick_sort_decrescente(vendas.copy())

print("Vendas mensais (original):", vendas)
print("Vendas ordenadas (decrescente):", vendas_ordenadas)
print("Melhor mês (maior venda):", vendas_ordenadas[0])
print("Pior mês (menor venda):", vendas_ordenadas[-1])
```

```python
#Projeto 2 — Jogo de Ordenação de Cartas

import random

cartas = [random.randint(1, 99) for _ in range(5)]

pilha_compras = cartas.copy()
random.shuffle(pilha_compras)

mao = []

print("Cartas embaralhadas na pilha:", pilha_compras)

while pilha_compras:
    input("\nPressione Enter para comprar uma carta...")

    carta = pilha_compras.pop()
    print(f"Você comprou a carta: {carta}")

    if not mao:
        mao.append(carta)
    else:
        pos = 0

        while pos < len(mao) and mao[pos] < carta:
            pos += 1

        mao.insert(pos, carta)

    print("Sua mão agora:", mao)

print("\nTodas as cartas foram compradas e ordenadas na mão!")

```
